# Image Classification with MobileNet V2

This notebook follows the provided TensorFlow tutorial using the pretrained `tf.keras.applications.MobileNetV2` model and the supplied highway image. It performs inference only; it does not train the model. Run each cell from top to bottom in Google Colab.


## 1. Install and import dependencies

TensorFlow includes Keras applications, including MobileNet V2.


```python
!pip -q install tensorflow matplotlib pillow

import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
from PIL import Image
from google.colab import files
from tensorflow.keras.applications.mobilenet_v2 import MobileNetV2, preprocess_input, decode_predictions
```

## 2. Upload and inspect the image

Upload `highway_sample.png`. The tutorial loads an image, displays it, and resizes it to the model input size.


```python
uploaded = files.upload()
IMAGE_PATH = next(iter(uploaded))
image = Image.open(IMAGE_PATH).convert("RGB")
print("Original image size:", image.size)
plt.figure(figsize=(9, 6))
plt.imshow(image)
plt.axis("off")
plt.title("Input image")
plt.show()
```

## 3. Prepare the input

MobileNet V2 expects a 224 × 224 RGB image. `preprocess_input` applies the preprocessing expected by this Keras model, and a batch dimension gives the input shape `(1, 224, 224, 3)`.


```python
IMAGE_SIZE = (224, 224)
resized = image.resize(IMAGE_SIZE)
image_array = np.asarray(resized, dtype=np.float32)
batch = np.expand_dims(image_array, axis=0)
batch = preprocess_input(batch)
print("Resized image shape:", image_array.shape)
print("Model input shape:", batch.shape)
```

## 4. Load MobileNet V2 and predict

The pretrained model uses ImageNet weights. `decode_predictions` converts the class scores into readable labels and returns the five highest scoring classes.


```python
model = MobileNetV2(weights="imagenet")
predictions = model.predict(batch, verbose=0)
top_predictions = decode_predictions(predictions, top=5)[0]

print("Top 5 predictions:")
for rank, (class_id, label, score) in enumerate(top_predictions, start=1):
    print(f"{rank}. {label} ({class_id}) — score {score:.4f}")
```

## 5. Display the top prediction

The plot shows the original image and the highest scoring prediction. Capture screenshots of the executed preprocessing, prediction output, and plot for the report.


```python
best_class_id, best_label, best_score = top_predictions[0]
plt.figure(figsize=(9, 6))
plt.imshow(image)
plt.axis("off")
plt.title(f"Top prediction: {best_label} ({best_score:.4f})")
plt.show()
```
